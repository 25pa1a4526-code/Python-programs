```python
#WRITE A PROGRAM TO CHECK IF A GIVEN KEY EXISTS IN A DICTIONARY OR NOT
d={'a':1,'b':2,'c':3,'d':4,'e':5}
key=input()
if key in d:
    print(key,"is present in the dictionary")
else:
    print(key,"is not present in the dictionary")
```

```python
#write a program to add a new key-value pair to an existing dictionary
d={'a':1,'b':2,'c':3,'d':4,'e':5}
key=input()
value=input()
d[key]=value
print(d)
```

```python
#write a program to sum all the items in a given dictionary
d={'a':10,'b':20,'c':30,'d':40,'e':50}
sum=0
for i in d.values():
    sum=sum+i
print(sum)
```

```python
#write a program that prints frequency of each element in alist in following
#format [1,2,3,4,5,1,5,6,1,6,7] {1:3,2:1,3:1,4:1,5:2,6:2,7:1}
list = [1,2,3,4,5,1,5,6,1,6,7]
freq = {}
for i in list:
  if i in freq:
    freq[i] += 1
  else:
    freq[i] = 1
print(freq)
```
