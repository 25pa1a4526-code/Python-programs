```python
#FILE HANDLING IN PYTHON
#PROGRAM ON "W" MODE
file=open("new.txt","w")
data="hello world"
file.write(data)
print("file created")
file.close()
```

```python
#FILE HANDLING IN PYTHON
file=open("f1.txt","w")
file.write("i created a file")
print("file is created")
file.close()
```

```python
#PROGRAM ON "R" MODE
file=open("new.txt","r")
data=file.read()
print(data)
file.close()

```

```python
#PROGRAM ON "a" MODE
file=open("new.txt","a")
data="hello world"
file.write(data)
print("file created")
file.close()
```

```python
with open("f1.txt","r") as f:
  data=f.read()
  print(data)
```

```python
with open("f1.txt","a") as f:
  f.writelines("\napple \ndrink \njuice")
  print("data is append")
```

```python
with open("f1.txt","r") as f:
  data=f.readlines()
  print(data)
```

```python
#COPYING CONTENT OF ONE FILE TO OTHER
with open("f1.txt","r") as f:
  data=f.read()
  print("reading done")
with open("f2.txt","w") as f2:
  f2.write(data)
  print("file copied")
```

```python
#WRITE A PROGRAM TO SORT WORDS IN A FILE
with open("f1.txt","r") as f:
  data=f.read().split()
  data_lower=[i.lower() for i in data]
  sort_data=sorted(data_lower)
with open("f4.txt","w") as f2:
  for i in sort_data:
    f2.write(i+"\n")
  print("done")
```

```python
#PYTHON PROGRAM TO COMPUTE THE NUMBER OF CHARACTERS,WORDS AND LINES IN A FILE
f=open("f1.txt","r")
data=f.read()
print("number of characters",len(data))
print("number of words",len(data.split()))
f.close
```

```python
import pickle
with open("f3.txt","wb") as f:
  data=[1,2,3,4,5,"hello"]
  pickle.dump(data,f)
with open("f3.txt","rb") as f:
  d=pickle.load(f)
  print(d)
```
